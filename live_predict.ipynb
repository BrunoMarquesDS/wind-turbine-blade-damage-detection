```python
import cv2
from ultralytics import YOLO

# Load the trained YOLOv8 model. The path 'weights/best.pt'
model = YOLO("weights/best.pt")

# Initialize video capture from the webcam.
cap = cv2.VideoCapture(1)

if not cap.isOpened():
    print("Could not open webcam.")
    exit()

# Create a full-screen window to display the real-time detections.
cv2.namedWindow("YOLOv8 - Real-Time Detection", cv2.WND_PROP_FULLSCREEN)
cv2.setWindowProperty("YOLOv8 - Real-Time Detection", cv2.WND_PROP_FULLSCREEN, cv2.WINDOW_FULLSCREEN)

print("Press 'q' to quit the live detection.")

# Start an infinite loop to process webcam frames in real time.
while True:
    ret, frame = cap.read()
    
    if not ret:
        print("Failed to grab frame.")
        break

    # Run object detection on the current frame using the YOLOv8 model.
    results = model(frame, verbose=False)

    # Iterate over each detection found in the frame.
    for result in results[0].boxes.data:
        x1, y1, x2, y2, conf, cls = result

        # Draw a rectangle (bounding box) on the image to highlight the detected object.
        cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (255, 0, 0), 2)

        label = f"Conf: {conf:.2f}"

        font_scale = 0.6
        font = cv2.FONT_HERSHEY_SIMPLEX
        color = (0, 255, 0)
        thickness = 1
        
        cv2.putText(frame, label, (int(x1), int(y1)-10), font, font_scale, color, thickness, cv2.LINE_AA)

    # Display the frame with the detections in the created window.
    cv2.imshow("YOLOv8 - Real-Time Detection", frame)

    # Check if the 'q' key was pressed. If so, break the loop.
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
