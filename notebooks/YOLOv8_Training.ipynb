# Wind Turbine Blade Damage Detection with YOLOv8

### 1. We install the ultralytics library to access the latest version of YOLOv8. Next, we import the YOLO class, which is the foundation for loading the model.

```python
from ultralytics import YOLO
```

### 2. We load the YOLOv8n model, which is the "nano" version—a lightweight, fast, and high-performance model, perfect for starting and testing our project. The .pt suffix indicates that it's a pre-trained weights file.

```python
model = YOLO("yolov8n.pt")
```

### 3. We use the model.train() method to start the model training. We pass the path to the dataset configuration file (data.yaml), define the number of epochs, and other settings like the batch size.

- data: Path to the dataset configuration file.

- epochs: Number of training epochs.

- batch: The number of images processed per iteration.

- imgsz: Resizes images to 640x640 pixels, a common standard for YOLO.

- optimizer: We use the SGD (Stochastic Gradient Descent) optimizer, which is effective and stable for this type of task.

- lr0: Sets the initial learning rate.

- warmup_epochs: Number of warmup epochs to gradually adjust the learning rate at the beginning.

- dropout: Dropout rate to prevent "overfitting".

- workers: Number of processes to load data.

- device: Defines the training device.

```python
model.train(
    data="path/data.yaml",
    epochs=150,
    batch=16,
    imgsz=640,
    optimizer="SGD",
    lr0=0.01,
    warmup_epochs=15,
    dropout=0.01,
    workers=4,
    device="cpu",
)
```

### 4. After training, we use the model.val() method to validate the model on the test dataset. This allows us to calculate performance metrics such as Precision, Recall, and mAP (mean Average Precision).

```python
model.val(data="path/data.yaml", split="test")
```
