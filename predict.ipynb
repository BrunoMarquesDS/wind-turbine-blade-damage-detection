### Video Test

```python
import cv2
from ultralytics import YOLO

# Load the trained model.
model = YOLO('weights/best.pt')

# Path to the video file.
video_path = 'videotest.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error opening video file")
    exit()

# Get the original video properties.
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Define the codec and create the VideoWriter to save the output video.
out = cv2.VideoWriter('video_with_detections.avi', cv2.VideoWriter_fourcc(*'XVID'), fps, (frame_width, frame_height))

# Process the video frame by frame.
while cap.isOpened():
    # Capture the next frame.
    ret, frame = cap.read()
    if not ret:
        break
    
    # Perform detection on the current frame.
    results = model(frame)
    
    # Access the first item in the results list.
    result = results[0]
    
    boxes = result.boxes.xyxy
    confidences = result.boxes.conf
    class_ids = result.boxes.cls
    
    # Draw the detection boxes on the frame.
    for i in range(len(boxes)):
        x1, y1, x2, y2 = boxes[i]
        confidence = confidences[i]
        class_id = class_ids[i]
        
        color = (0, 255, 0)
        
        # Draw the bounding box.
        cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), color, 2)
        
        label = f'{model.names[int(class_id)]}: {confidence:.2f}'
        cv2.putText(frame, label, (int(x1), int(y1) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

    out.write(frame)

cap.release()
out.release()
cv2.destroyAllWindows()

print("Video saved as 'video_with_detections.avi'")
```
